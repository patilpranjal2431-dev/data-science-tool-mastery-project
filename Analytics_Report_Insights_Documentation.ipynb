# Analytics Report & Insights Documentation

## Stakeholder Analytics Case Study — Retail Performance

### Business scenario
A retail business wants a management-level view of sales performance across regions, categories, channels, and customer types. Leadership needs to understand revenue growth, profitability, customer satisfaction, and returns so that resources can be prioritized.

### Core business questions
1. How is revenue performing over time?
2. Which regions and categories contribute most to revenue and profit?
3. Which channel performs better?
4. Where are returns or satisfaction creating attention areas?
5. What actions should management prioritize?

**Important:** The dataset is synthetic and created for an educational analytics case study.

## 1. Business Objectives and KPI Definitions

**Primary objective:** improve profitable revenue while protecting customer experience.

### KPIs
- Total Revenue
- Total Profit
- Profit Margin
- Order Count
- Average Order Value
- Return Rate
- Average Satisfaction

### Stakeholders
- Executive leadership: overall performance and priorities
- Sales managers: regional and channel performance
- Category managers: category profitability
- Customer-experience teams: satisfaction and returns

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv("data/retail_orders.csv",parse_dates=["order_date"])
df["month"]=df.order_date.dt.to_period("M").astype(str)
print(df.shape)
display(df.head())
display(df.describe(include="all").T)
```

## 2. Data Quality and Preparation

```python
print("Missing values:")
display(df.isna().sum().to_frame("missing"))
print("Duplicate rows:",df.duplicated().sum())

# Basic validation
assert (df.revenue >= 0).all()
assert (df.profit <= df.revenue).all()
assert df.discount_pct.between(0,35).all()
assert df.satisfaction.between(1,5).all()

print("Data validation checks passed.")
```

## 3. Executive KPI Snapshot

```python
kpis=pd.Series({
"Total Revenue":df.revenue.sum(),
"Total Profit":df.profit.sum(),
"Profit Margin":df.profit.sum()/df.revenue.sum(),
"Orders":len(df),
"Average Order Value":df.revenue.mean(),
"Return Rate":df.returned.mean(),
"Average Satisfaction":df.satisfaction.mean()
})
display(kpis.to_frame("value"))
```

## 4. Revenue Trend

The trend is examined at monthly level to identify growth, decline, or unusual periods.

```python
monthly=df.groupby("month").agg(revenue=("revenue","sum"),profit=("profit","sum"),orders=("order_id","count")).reset_index()
display(monthly)
plt.figure(figsize=(10,5)); plt.plot(monthly.month,monthly.revenue,marker="o"); plt.xticks(rotation=45); plt.title("Monthly Revenue Trend"); plt.xlabel("Month"); plt.ylabel("Revenue"); plt.tight_layout(); plt.show()
```

## 5. Regional Performance

```python
region=df.groupby("region").agg(revenue=("revenue","sum"),profit=("profit","sum"),orders=("order_id","count"),satisfaction=("satisfaction","mean"),return_rate=("returned","mean")).sort_values("revenue",ascending=False)
display(region)
plt.figure(figsize=(8,5)); plt.bar(region.index,region.revenue); plt.title("Revenue by Region"); plt.xlabel("Region"); plt.ylabel("Revenue"); plt.tight_layout(); plt.show()
```

## 6. Category Performance

```python
category=df.groupby("category").agg(revenue=("revenue","sum"),profit=("profit","sum"),orders=("order_id","count"),return_rate=("returned","mean")).sort_values("revenue",ascending=False)
display(category)
plt.figure(figsize=(8,5)); plt.bar(category.index,category.profit); plt.title("Profit by Category"); plt.xlabel("Category"); plt.ylabel("Profit"); plt.tight_layout(); plt.show()
```

## 7. Channel Performance

```python
channel=df.groupby("channel").agg(revenue=("revenue","sum"),profit=("profit","sum"),orders=("order_id","count"),return_rate=("returned","mean")).sort_values("revenue",ascending=False)
display(channel)
plt.figure(figsize=(8,5)); plt.bar(channel.index,channel.revenue); plt.title("Revenue by Channel"); plt.xlabel("Channel"); plt.ylabel("Revenue"); plt.tight_layout(); plt.show()
```

## 8. Customer Experience and Returns

Returns and satisfaction are reviewed alongside commercial metrics. A high-revenue segment can still require intervention if customer-experience indicators are weak.

```python
experience=df.groupby(["category","channel"]).agg(
orders=("order_id","count"),revenue=("revenue","sum"),profit=("profit","sum"),
satisfaction=("satisfaction","mean"),return_rate=("returned","mean")
).sort_values("return_rate",ascending=False)
display(experience.head(10))
```

## 9. Key Findings

The analysis should distinguish **what happened**, **why it may matter**, and **what management can do next**.

### Findings
- Revenue performance should be monitored by month rather than relying only on the total.
- Regional contribution is uneven, creating opportunities for targeted sales investment.
- Category revenue and category profit are not necessarily ranked identically because margins differ.
- Online and store channels should be evaluated using both revenue and customer-experience metrics.
- Return rate is particularly useful for identifying categories or channel combinations requiring operational review.
- Satisfaction should be tracked together with commercial KPIs because aggressive discounting or volume growth can create downstream customer-experience effects.

### Responsible interpretation
These findings are descriptive. The dataset does not establish causal relationships. Management decisions should be validated against operational, pricing, inventory, and customer-service information.

## 10. Actionable Recommendations

### Priority 1 — Protect profitable growth
Allocate additional sales attention to high-profit regions and categories, not simply the highest-revenue segments.

### Priority 2 — Investigate high-return segments
Review product quality, descriptions, sizing/fit information, fulfillment, and return reasons for high-return category/channel combinations.

### Priority 3 — Improve channel decisions
Compare online and store performance using revenue, margin, return rate, and satisfaction together rather than using revenue alone.

### Priority 4 — Establish a recurring KPI dashboard
Refresh revenue, profit margin, orders, AOV, return rate, and satisfaction monthly with region/category/channel drill-downs.

### Priority 5 — Create an action-owner framework
Assign owners and deadlines to each underperforming region, category, or customer-experience metric.

## 11. Executive Reporting Template

A stakeholder-facing report should answer:

**What happened?** Revenue, profit, and KPI trend.

**Where did it happen?** Region, category, and channel.

**Why does it matter?** Financial impact, customer impact, and operational risk.

**What should we do?** Prioritized actions with owners and next steps.

**How will we know it worked?** Define measurable KPI targets and review cadence.
